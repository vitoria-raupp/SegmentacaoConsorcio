```python
import pandas as pd
import numpy as np

# ============================================================
# 1. PREPARAÇÃO DA BASE
# ============================================================

df_modelo = df.copy()

df_modelo["dt_ultima_compra"] = pd.to_datetime(
    df_modelo["dt_ultima_compra"]
)

# ------------------------------------------------------------
# Padronização de texto
# ------------------------------------------------------------

df_modelo["tipo_pessoa"] = (
    df_modelo["tipo_pessoa"]
    .astype(str)
    .str.strip()
    .str.upper()
)

df_modelo["produto"] = (
    df_modelo["produto"]
    .astype(str)
    .str.strip()
    .str.upper()
)

df_modelo["situacao_adimplencia"] = (
    df_modelo["situacao_adimplencia"]
    .astype(str)
    .str.strip()
    .str.upper()
)

df_modelo["teve_cotas_canceladas"] = (
    df_modelo["teve_cotas_canceladas"]
    .astype(str)
    .str.strip()
    .str.upper()
)


# ============================================================
# 2. FLAGS AUXILIARES
# ============================================================

df_modelo["inadimplente_flag"] = (
    df_modelo["situacao_adimplencia"]
    .eq("INADIMPLENTE")
)

df_modelo["cancelamento_flag"] = (
    df_modelo["teve_cotas_canceladas"]
    .eq("SIM")
)

df_modelo["tem_veiculos_row"] = (
    df_modelo["produto"]
    .str.contains(
        "VEÍC|VEIC",
        regex=True,
        na=False
    )
)

df_modelo["tem_imoveis_row"] = (
    df_modelo["produto"]
    .str.contains(
        "IMÓV|IMOV",
        regex=True,
        na=False
    )
)


# ============================================================
# 3. MANTER TODOS OS REGISTROS SEPARADOS
# ============================================================

# NÃO fazemos groupby.
# Cada linha da base original permanece uma linha.

clientes = df_modelo.copy()

print("Registros originais:", len(df))
print("Registros na base de score:", len(clientes))

# Os valores devem ser iguais.
```

```python
# ============================================================
# 4. RECÊNCIA
# ============================================================

data_referencia = clientes["dt_ultima_compra"].max()

clientes["recencia_meses"] = (
    (
        data_referencia
        - clientes["dt_ultima_compra"]
    ).dt.days
    / 30.44
).round(1)

clientes["recencia_meses"] = (
    clientes["recencia_meses"]
    .clip(lower=0)
)
```

```python
# ============================================================
# 4. RECÊNCIA
# ============================================================

data_referencia = clientes["dt_ultima_compra"].max()

clientes["recencia_meses"] = (
    (
        data_referencia
        - clientes["dt_ultima_compra"]
    ).dt.days
    / 30.44
).round(1)

clientes["recencia_meses"] = (
    clientes["recencia_meses"]
    .clip(lower=0)
)
```

```python
# ============================================================
# 5. ELEGIBILIDADE
# ============================================================

clientes["elegivel"] = (
    (~clientes["inadimplente_flag"])
    &
    (~clientes["cancelamento_flag"])
    &
    (clientes["vl_renda_mensal"] > 0)
    &
    (clientes["recencia_meses"] <= 24)
)
```

```python
# ============================================================
# 6. SCORE DE CAPACIDADE FINANCEIRA - VEÍCULOS
# ============================================================

clientes["score_capacidade_veiculos"] = np.select(
    [
        clientes["vl_renda_mensal"] >= 20000,
        clientes["vl_renda_mensal"] >= 15000,
        clientes["vl_renda_mensal"] >= 10000,
        clientes["vl_renda_mensal"] >= 5000
    ],
    [
        30,
        25,
        20,
        10
    ],
    default=5
)
```

```python
# ============================================================
# 6. SCORE DE CAPACIDADE FINANCEIRA - VEÍCULOS
# ============================================================

clientes["score_capacidade_veiculos"] = np.select(
    [
        clientes["vl_renda_mensal"] >= 20000,
        clientes["vl_renda_mensal"] >= 15000,
        clientes["vl_renda_mensal"] >= 10000,
        clientes["vl_renda_mensal"] >= 5000
    ],
    [
        30,
        25,
        20,
        10
    ],
    default=5
)
```

```python
# ============================================================
# 7. SCORE DE CAPACIDADE FINANCEIRA - IMÓVEIS
# ============================================================

clientes["score_capacidade_imoveis"] = np.select(
    [
        clientes["vl_renda_mensal"] >= 40000,
        clientes["vl_renda_mensal"] >= 30000,
        clientes["vl_renda_mensal"] >= 20000,
        clientes["vl_renda_mensal"] >= 10000
    ],
    [
        30,
        25,
        20,
        10
    ],
    default=5
)
```

```python
# ============================================================
# 8. SCORE DE RECÊNCIA
# ============================================================

clientes["score_recencia"] = np.select(
    [
        clientes["recencia_meses"] <= 6,
        clientes["recencia_meses"] <= 12,
        clientes["recencia_meses"] <= 24
    ],
    [
        20,
        15,
        10
    ],
    default=0
)
```

```python
# ============================================================
# 9. SCORE DE ENGAJAMENTO
# ============================================================

clientes["score_engajamento"] = np.select(
    [
        clientes["qtd_cotas_total"] >= 5,
        clientes["qtd_cotas_total"] >= 3,
        clientes["qtd_cotas_total"] >= 2
    ],
    [
        20,
        15,
        10
    ],
    default=5
)
```

```python
# ============================================================
# 10. SCORE DE RELACIONAMENTO
# ============================================================

clientes["score_relacionamento"] = np.select(
    [
        clientes["tempo_relacionamento_anos"] >= 5,
        clientes["tempo_relacionamento_anos"] >= 3,
        clientes["tempo_relacionamento_anos"] >= 1
    ],
    [
        15,
        12,
        8
    ],
    default=5
)
```

```python
# ============================================================
# 11. SCORE DE AFINIDADE COM PRODUTO
# ============================================================

clientes["score_afinidade_veiculos"] = np.where(
    clientes["tem_veiculos_row"],
    15,
    10
)

clientes["score_afinidade_imoveis"] = np.where(
    clientes["tem_imoveis_row"],
    15,
    10
)
```

```python
# ============================================================
# 12. SCORE FINAL - VEÍCULOS
# ============================================================

clientes["score_veiculos"] = (
    clientes["score_capacidade_veiculos"]
    + clientes["score_recencia"]
    + clientes["score_engajamento"]
    + clientes["score_relacionamento"]
    + clientes["score_afinidade_veiculos"]
)


# ============================================================
# 13. SCORE FINAL - IMÓVEIS
# ============================================================

clientes["score_imoveis"] = (
    clientes["score_capacidade_imoveis"]
    + clientes["score_recencia"]
    + clientes["score_engajamento"]
    + clientes["score_relacionamento"]
    + clientes["score_afinidade_imoveis"]
)
```

```python
# ============================================================
# 14. PRODUTO RECOMENDADO
# ============================================================

clientes["produto_recomendado"] = np.select(
    [
        ~clientes["elegivel"],

        clientes["score_veiculos"]
        > clientes["score_imoveis"],

        clientes["score_imoveis"]
        > clientes["score_veiculos"]
    ],
    [
        "Não ofertar",
        "VEÍCULOS",
        "IMÓVEIS"
    ],
    default="AMBOS"
)
```

```python
# ============================================================
# 15. SCORE GERAL CONQUISTE+
# ============================================================

clientes["score_conquiste"] = clientes[
    [
        "score_veiculos",
        "score_imoveis"
    ]
].max(axis=1)
```

```python
# ============================================================
# 16. SEGMENTAÇÃO FINAL
# ============================================================

clientes["segmento_conquiste"] = np.select(
    [
        ~clientes["elegivel"],
        clientes["score_conquiste"] >= 80,
        clientes["score_conquiste"] >= 65
    ],
    [
        "Não prioritário",
        "Alta prioridade",
        "Potencial"
    ],
    default="Nutrição"
)
```

```python
# ============================================================
# 17. SCORE DO PRODUTO ATUAL DA LINHA
# ============================================================

clientes["score_produto_atual"] = np.select(
    [
        clientes["tem_veiculos_row"],
        clientes["tem_imoveis_row"]
    ],
    [
        clientes["score_veiculos"],
        clientes["score_imoveis"]
    ],
    default=np.nan
)
```

```python
# ============================================================
# 18. SEGMENTO DO PRODUTO ATUAL
# ============================================================

clientes["segmento_produto_atual"] = np.select(
    [
        ~clientes["elegivel"],
        clientes["score_produto_atual"] >= 80,
        clientes["score_produto_atual"] >= 65
    ],
    [
        "Não prioritário",
        "Alta prioridade",
        "Potencial"
    ],
    default="Nutrição"
)
```
